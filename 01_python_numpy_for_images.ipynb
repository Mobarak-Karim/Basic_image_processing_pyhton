## Images are arrays

A grayscale image is commonly a 2-D array: rows × columns. A color image usually has a third axis for channels.

```python
import numpy as np

tiny = np.array([[0,25,50],[75,100,125],[150,200,255]])
print(tiny)
print('shape:', tiny.shape)
print('dtype:', tiny.dtype)
```

## Indexing: read one pixel or a region

```python
# NumPy uses [row, column], not [x, y].
print('center pixel:', tiny[1,1])

# Slice rows 0..1 and columns 1..2.
crop = tiny[0:2,1:3]
print(crop)
```

## Vectorized operations

```python
# NumPy can apply an operation to the full array at once.
bright = np.clip(tiny + 30, 0, 255)
print(bright)

# Boolean masks are used constantly in image analysis.
mask = tiny > 100
print(mask)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski

def show_image(image, title='', cmap=None, figsize=(6, 5)):
    """Display one image with a clean, repeatable layout."""
    plt.figure(figsize=figsize)
    plt.imshow(image, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
image = ski.data.moon()
show_image(image, 'Original moon image', cmap='gray')

crop = image[150:350,120:380]
show_image(crop, 'Cropped region', cmap='gray')
print('mean intensity:', crop.mean())
print('standard deviation:', crop.std())
```
