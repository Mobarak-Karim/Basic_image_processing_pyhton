## 1. What Python code is doing

Python executes instructions from top to bottom.

A line such as:

```python
pixel_value = 125
```

means:

1. evaluate the value on the right (`125`),
2. store it in a variable named `pixel_value`.

A **variable** is simply a name that refers to a value.

````python
# A variable gives a useful name to a value.
pixel_value = 125

# print(...) is a FUNCTION CALL.
# We pass pixel_value into print as an argument.
print(pixel_value)
````

### Why variables matter in image processing

Good variable names make processing steps readable:

```python
raw_image = ...
smoothed_image = ...
threshold = ...
binary_mask = ...
labels = ...
```

That is much easier to understand than repeatedly overwriting a variable called `x`.

## 3. Operators: calculations and comparisons

Arithmetic operators:

```text
+   add
-   subtract
*   multiply
/   divide
**  exponent
%   remainder
```

Comparison operators return a Boolean value:

```text
>   greater than
<   less than
>=  greater than or equal
<=  less than or equal
==  equal
!=  not equal
```

**Important:** `=` assigns a value. `==` compares two values.

````python
a = 10
b = 4

print("a + b =", a + b)
print("a / b =", a / b)
print("a ** 2 =", a ** 2)

print("Is a greater than b?", a > b)
print("Is a equal to b?", a == b)
````

### Indexing and slicing

## 6. Logic with `if`, `elif`, and `else`

Use conditionals when the program should make a decision.

```python
if condition:
    ...
elif another_condition:
    ...
else:
    ...
```

Python uses **indentation** to define the code that belongs to each branch.

````python
mean_intensity = 145

if mean_intensity > 180:
    print("Image is relatively bright.")
elif mean_intensity > 100:
    print("Image has intermediate brightness.")
else:
    print("Image is relatively dark.")
````

### Boolean logic: `and`, `or`, `not`

Use:

- `and` when **both** conditions must be true,
- `or` when **at least one** condition must be true,
- `not` to reverse a Boolean value.

````python
area = 250
mean_intensity = 160

keep_object = (area > 100) and (mean_intensity > 120)

print("keep object?", keep_object)
print("reject object?", not keep_object)
````

## 7. `for` loops

A `for` loop repeats a block of code for each item in a collection.

This is very useful for:

- processing several files,
- testing several parameter values,
- iterating over measurements.

Use loops when the *workflow* must repeat. Later, NumPy will let us avoid loops for many pixel-by-pixel operations.

````python
filenames = ["img_01.tif", "img_02.tif", "img_03.tif"]

for filename in filenames:
    print("Processing:", filename)
````

### `range(...)` and `enumerate(...)`

`range(...)` generates integer sequences.

`enumerate(...)` gives both an index and the current item.

````python
for i in range(3):
    print("range value:", i)

print("---")

for index, filename in enumerate(filenames):
    print(index, filename)
````

## 8. `while` loops

A `while` loop repeats as long as a condition stays true.

Use it when you do **not** know the number of repetitions in advance.

Be careful: if the condition never becomes false, the loop will run forever.

````python
count = 0

while count < 3:
    print("count =", count)
    count = count + 1  # update the value so the loop can eventually stop
````

## 9. Calling functions

A function is reusable code that performs a task.

General form:

```python
result = function_name(argument1, argument2, keyword=value)
```

Examples:

```python
print("hello")
len(filenames)
round(3.14159, 2)
```

The parentheses `()` mean **call the function**.

````python
print("Number of files:", len(filenames))
print("Rounded value:", round(3.14159, 2))
````

### Positional arguments vs keyword arguments

A **positional argument** is interpreted by its position.

A **keyword argument** explicitly names the parameter.

Keyword arguments often make scientific code easier to understand:

```python
filters.gaussian(image, sigma=2.0)
```

The reader can immediately see that `2.0` controls `sigma`.

## 10. Writing your own function with `def`

Functions are one of the most important tools for reproducible image analysis.

Use a function when:

- the same operation appears more than once,
- you want to test the same workflow on several images,
- you want parameters to be explicit,
- you want one logical task to have one clear name.

````python
def classify_intensity(value, threshold=100):
    '''
    Return a text label based on an intensity threshold.

    Parameters
    ----------
    value : float
        Intensity value to classify.
    threshold : float, optional
        Boundary between "low" and "high".
    '''
    if value > threshold:
        return "high"
    else:
        return "low"


# CALL the function:
label_1 = classify_intensity(145)
label_2 = classify_intensity(80, threshold=90)

print(label_1)
print(label_2)
````

### Parameters, arguments, and `return`

In:

```python
def classify_intensity(value, threshold=100):
```

`value` and `threshold` are **parameters**.

In:

```python
classify_intensity(145, threshold=120)
```

`145` and `120` are **arguments**.

`return` sends a result back to the caller.

A function without `return` can still perform an action, but it does not return a useful result by default.

## 12. Errors and a practical debugging habit

Errors are normal.

Common beginner errors include:

- `NameError` — variable/function name does not exist,
- `TypeError` — operation used with an incompatible type,
- `IndexError` — index outside the valid range,
- `KeyError` — dictionary key does not exist,
- `IndentationError` — indentation is inconsistent,
- `ModuleNotFoundError` — package is not installed in the active environment.

### Debugging habit

When something fails:

1. read the **last line** of the traceback first,
2. identify the exception type,
3. read the error message,
4. locate the line in *your code*,
5. inspect the variables immediately before that line,
6. change one thing and rerun.

Do not randomly change many lines at once.

````python
# A safe debugging example:
values = [10, 20, 30]

print("list length:", len(values))
print("valid indices are 0 through", len(values) - 1)

# values[3] would raise IndexError because the last valid index is 2.
````

# Part II — NumPy and image arrays

Now we connect Python fundamentals to image processing.

The key idea is:

> **A digital image is numerical data arranged in an array.**

For a grayscale image:

```text
rows × columns
```

For an RGB image:

```text
rows × columns × channels
```

````python
# Create a tiny 3x3 grayscale image by hand.
tiny = np.array([
    [0, 25, 50],
    [75, 100, 125],
    [150, 200, 255],
], dtype=np.uint8)

print(tiny)
print("shape:", tiny.shape)
print("dtype:", tiny.dtype)
print("dimensions:", tiny.ndim)
````

## 13. NumPy indexing uses `[row, column]`

This is extremely important.

```python
image[row, column]
```

not:

```python
image[x, y]
```

For image coordinates, it is often helpful to think:

```text
row    = vertical position
column = horizontal position
```

````python
# Center pixel: second row, second column.
print("center pixel:", tiny[1, 1])

# Crop rows 0-1 and columns 1-2.
crop = tiny[0:2, 1:3]
print(crop)
````

## 14. NumPy statistics

Common functions and methods:

```python
image.min()
image.max()
image.mean()
image.std()
np.median(image)
```

Use these to inspect numerical data. They do not replace visualization, but they quickly reveal unexpected ranges or values.

````python
print("minimum:", tiny.min())
print("maximum:", tiny.max())
print("mean   :", tiny.mean())
print("median :", np.median(tiny))
print("std    :", tiny.std())
````

## 15. Vectorized operations

NumPy can apply an operation to an entire array at once.

This is usually clearer and faster than writing a Python loop over individual pixels.

### Important dtype warning

Unsigned integer image types such as `uint8` can overflow during arithmetic. Convert to a suitable working type or use image-aware conversion before quantitative processing.

````python
# Convert to float before arithmetic so values do not wrap around.
tiny_float = tiny.astype(float)

brighter = np.clip(tiny_float + 30, 0, 255)

print(brighter)
````

## 16. Boolean masks

A comparison applied to an array produces a Boolean array.

This is the foundation of threshold-based segmentation.

```python
mask = image > threshold
```

Each pixel becomes:

- `True` if the condition is satisfied,
- `False` otherwise.

````python
mask = tiny > 100

print(mask)
print("number of True pixels:", mask.sum())
````

## 17. A real image

Now use the same array concepts on an actual image from `skimage.data`.

````python
import matplotlib.pyplot as plt
import skimage as ski

image = ski.data.moon()

print("shape:", image.shape)
print("dtype:", image.dtype)
print("range:", image.min(), "to", image.max())

plt.figure(figsize=(6, 5))
plt.imshow(image, cmap="gray")
plt.title("Moon image")
plt.axis("off")
plt.show()
````

## 18. Cropping a real image

Cropping is still ordinary NumPy slicing.

Always remember:

```python
image[row_start:row_end, column_start:column_end]
```

````python
crop = image[150:350, 120:380]

plt.figure(figsize=(6, 5))
plt.imshow(crop, cmap="gray")
plt.title("Cropped region")
plt.axis("off")
plt.show()

print("crop shape:", crop.shape)
print("mean intensity:", crop.mean())
print("standard deviation:", crop.std())
````

## 19. Connect Python logic to image analysis

The same Python concepts now become image-processing tools:

| Python concept | Image-processing use |
|---|---|
| variable | store an image, mask, threshold, measurement |
| comparison | create a Boolean mask |
| `if` | make a decision based on metadata or measurements |
| `for` loop | process multiple files |
| function | make a reusable analysis step |
| list | store filenames or results |
| dictionary | store parameters / metadata |
| NumPy array | represent image pixels |

This connection is more important than memorizing individual commands.

````python
def summarize_image(image):
    '''
    Return a dictionary with basic image statistics.

    This is a realistic example of combining a function,
    NumPy-style array methods, and a dictionary.
    '''
    summary = {
        "shape": image.shape,
        "dtype": str(image.dtype),
        "min": float(image.min()),
        "max": float(image.max()),
        "mean": float(image.mean()),
    }
    return summary


summary = summarize_image(image)

for key, value in summary.items():
    print(f"{key}: {value}")
````
