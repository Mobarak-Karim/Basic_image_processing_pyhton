## Color images are multichannel arrays

An RGB image usually has shape `(rows, columns, 3)`. Biomedical images may contain multiple fluorescence channels that are not stored as RGB.

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski

def show_image(image, title='', cmap=None, figsize=(6, 5)):
    """Display one image with a clean, repeatable layout."""
    plt.figure(figsize=figsize)
    plt.imshow(image, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
image = ski.data.immunohistochemistry()
print('shape:', image.shape)
print('dtype:', image.dtype)
show_image(image, 'Immunohistochemistry example')
```

## Inspect channels separately

```python
channel_names=['Red','Green','Blue']
fig,axes=plt.subplots(1,3,figsize=(13,4))
for idx,ax in enumerate(axes):
    ax.imshow(image[...,idx],cmap='gray'); ax.set_title(channel_names[idx]); ax.axis('off')
plt.tight_layout(); plt.show()
```

## Convert RGB to grayscale

```python
from skimage import color
gray = color.rgb2gray(image)
show_image(gray, 'Grayscale conversion', cmap='gray')
print('range:', gray.min(), 'to', gray.max())
```

## Synthetic two-channel fluorescence example

This image is generated in code, so the course does not depend on external files.

```python
from skimage.draw import disk
from skimage.filters import gaussian

rng=np.random.default_rng(7)
shape=(256,256)
nuclei=np.zeros(shape,float); cytoplasm=np.zeros(shape,float)
centers=[(65,70),(85,170),(160,85),(175,175)]
for r,c in centers:
    rr,cc=disk((r,c),14,shape=shape); nuclei[rr,cc]=1.0
    rr,cc=disk((r,c),28,shape=shape); cytoplasm[rr,cc]+=0.55
nuclei=gaussian(nuclei,sigma=2)+rng.normal(0,0.03,shape)
cytoplasm=gaussian(cytoplasm,sigma=4)+rng.normal(0,0.03,shape)
composite=np.zeros((*shape,3),float)
composite[...,1]=np.clip(cytoplasm,0,1)
composite[...,2]=np.clip(nuclei,0,1)
show_image(composite, 'Synthetic two-channel fluorescence image')
```
