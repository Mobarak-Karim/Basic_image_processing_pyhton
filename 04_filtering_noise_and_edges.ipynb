## Filtering has a purpose

Decide what you are trying to reduce or emphasize: random noise, isolated outliers, large background variation, or edges.

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski

def show_image(image, title='', cmap=None, figsize=(6, 5)):
    """Display one image with a clean, repeatable layout."""
    plt.figure(figsize=figsize)
    plt.imshow(image, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
from skimage import util, filters
clean = ski.data.camera()
gaussian_noise = util.random_noise(clean, mode='gaussian', var=0.01, rng=42)
salt_pepper = util.random_noise(clean, mode='s&p', amount=0.03, rng=42)
```

## Gaussian and median filters

```python
gaussian_filtered = filters.gaussian(gaussian_noise, sigma=1.2)
median_filtered = filters.median(salt_pepper, footprint=np.ones((3,3)))
fig,axes=plt.subplots(2,2,figsize=(10,9))
for ax,img,title in zip(axes.ravel(),[gaussian_noise,gaussian_filtered,salt_pepper,median_filtered],['Gaussian noise','Gaussian filter','Salt & pepper noise','Median filter']):
    ax.imshow(img,cmap='gray'); ax.set_title(title); ax.axis('off')
plt.tight_layout(); plt.show()
```

## Edge detection

```python
edge_image = ski.data.coins()
sobel_edges = filters.sobel(edge_image)
fig,axes=plt.subplots(1,2,figsize=(10,4))
axes[0].imshow(edge_image,cmap='gray'); axes[0].set_title('Original')
axes[1].imshow(sobel_edges,cmap='magma'); axes[1].set_title('Sobel edge magnitude')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()
```

## Choosing a filter
- **Gaussian:** smooths high-frequency variation.
- **Median:** useful for impulsive noise while preserving edges better.
- **Sobel:** estimates edge strength rather than denoising.
