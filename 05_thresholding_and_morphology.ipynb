## Thresholding creates a first mask

Thresholding converts a grayscale image into a Boolean image. It works best when object and background intensities differ.

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski

def show_image(image, title='', cmap=None, figsize=(6, 5)):
    """Display one image with a clean, repeatable layout."""
    plt.figure(figsize=figsize)
    plt.imshow(image, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
from skimage import filters, morphology
image = ski.data.coins()
threshold = filters.threshold_otsu(image)
mask = image > threshold
print('Otsu threshold:', threshold)
fig,axes=plt.subplots(1,2,figsize=(10,4))
axes[0].imshow(image,cmap='gray'); axes[0].set_title('Input')
axes[1].imshow(mask,cmap='gray'); axes[1].set_title('Otsu mask')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()
```

## Clean the mask with morphology

```python
cleaned = morphology.remove_small_objects(mask, max_size=79)
cleaned = morphology.remove_small_holes(cleaned, max_size=79)
fig,axes=plt.subplots(1,2,figsize=(10,4))
axes[0].imshow(mask,cmap='gray'); axes[0].set_title('Raw mask')
axes[1].imshow(cleaned,cmap='gray'); axes[1].set_title('Cleaned mask')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()
```

## Global versus local thresholding

```python
uneven = ski.data.page()
global_t = filters.threshold_otsu(uneven)
global_mask = uneven > global_t
local_t = filters.threshold_local(uneven, block_size=51, offset=5)
local_mask = uneven > local_t
fig,axes=plt.subplots(1,3,figsize=(14,4))
for ax,img,title in zip(axes,[uneven,global_mask,local_mask],['Uneven illumination','Global threshold','Local threshold']):
    ax.imshow(img,cmap='gray'); ax.set_title(title); ax.axis('off')
plt.tight_layout(); plt.show()
```
