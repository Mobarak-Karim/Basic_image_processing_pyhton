# Part D — How to translate an equation into Python

This is a core scientific-programming skill.

Suppose

\[ A = \pi r^2 \]

### Step 1 — Identify inputs
Radius `r`.

### Step 2 — Identify output
Area `A`.

### Step 3 — Translate symbols
- multiplication → `*`
- exponent → `**`
- π → `math.pi` or `np.pi`

### Step 4 — Preserve mathematical grouping
Use parentheses where helpful.

### Step 5 — Track units
If radius is µm, area is µm².

### Step 6 — Test a known case
For radius 1, the answer should be approximately π.

### Formula-to-code checklist
1. Write the equation clearly.
2. List every symbol.
3. State units.
4. Decide inputs.
5. Decide output.
6. Translate operators.
7. Preserve grouping with parentheses.
8. Test a simple case.
9. Check expected direction/trend.
10. Check output units.
11. Only then apply to real data.

````python
import math
radius_um = 5
area_um2 = math.pi * radius_um**2
print('Area =', area_um2, 'µm²')
````

## Example — exponential attenuation / Beer–Lambert style model

\[ I = I_0 e^{-\mu z} \]

Ask before coding:
- What are the inputs?
- What are their units?
- Is the exponent dimensionless?
- Should intensity decrease as depth increases?
- At `z = 0`, should `I == I0`?

Scientific code should reproduce these expected behaviors.

````python
I0 = 1.0
mu_per_mm = 2.0
for z_mm in [0.0, 0.5, 1.0, 2.0]:
    I = I0 * math.exp(-mu_per_mm * z_mm)
    print(f'z={z_mm:.1f} mm -> I={I:.4f}')
````

## `match / case`

Python 3.10+ provides `match / case` for named categories or modes. Use it when one variable can take one of several specific values. For numeric thresholds, `if / elif / else` is usually clearer. The `_` case means anything else.

````python
method = 'otsu'
match method:
    case 'otsu':
        message = 'Use a global Otsu threshold.'
    case 'local':
        message = 'Use a local threshold.'
    case _:
        message = 'Unknown method.'
print(message)
````

# Part J — Functions

A function is reusable logic. Calling a function uses parentheses.

Built-in functions: `print()`, `len()`, `type()`

Library functions: `np.mean()`, `plt.imshow()`

When you define your own function:

```python
def function_name(parameters):
    result = ...
    return result
```

A **parameter** is the name in the definition. An **argument** is the actual value supplied when calling the function.

````python
def circle_area(radius):
    area = math.pi * radius**2
    return area

result = circle_area(5)
print(result)
````

## Default and keyword arguments

Defaults make parameters optional. Keyword arguments make calls easier to read.

````python
def linear_model(x, slope=1.0, intercept=0.0):
    return slope * x + intercept

print(linear_model(4))
print(linear_model(4, slope=2.5))
print(linear_model(4, slope=2.5, intercept=1.2))
````

## `return` versus `print`

`print()` displays a value. `return` sends a value back to the caller so it can be stored, combined, tested, or passed into another function. Scientific analysis functions should generally return useful results.

````python
def summarize(values):
    minimum = min(values)
    maximum = max(values)
    mean = sum(values) / len(values)
    return minimum, maximum, mean

minimum, maximum, mean = summarize([2, 4, 8, 10])
print(minimum, maximum, mean)
````

## Scope and docstrings

Variables created inside a function are normally local. This helps prevent accidental interference. A docstring should explain the function purpose, parameters, units, and return values when relevant.

````python
def attenuation(I0, mu_per_mm, z_mm):
    """Return I0*exp(-mu*z); mu in 1/mm and z in mm."""
    return I0 * math.exp(-mu_per_mm * z_mm)

print(attenuation(1.0, 2.0, 0.5))
````

# Part M — How to conceptualize a coding problem

Do **not** start with code. Start with the question.

Example:
> Measure mean intensity of pixels brighter than a threshold.

Plain-language algorithm:
1. Load image.
2. Choose threshold.
3. Compare each pixel with threshold.
4. Keep pixels above threshold.
5. Calculate their mean.

Then write pseudocode, then Python.

### Reusable scientific problem-solving template
A. Define the question.
B. Define the input.
C. Define the output.
D. Write the formula or rule.
E. Break it into small steps.
F. Test with tiny values you can verify manually.
G. Convert working logic into a function.
H. Apply to real data.
I. Validate visually/numerically.

# Part N — More formula translation examples

## Euclidean distance
\[ d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2} \]

Breaking a long expression into meaningful intermediate variables makes it easier to verify.

## Min–max normalization
\[x_{norm}=\frac{x-x_{min}}{x_{max}-x_{min}}\]

Before coding ask what happens when `x_max == x_min`.

## Gaussian
\[G(x)=A\exp[-(x-\mu)^2/(2\sigma^2)]\]

Translate one conceptual piece at a time.

## Units
For square pixels:
\[A_{\mu m^2}=A_{pixels}(pixel\ size)^2\]

Keep units visible in variable names whenever practical.

````python
# Euclidean distance
x1, y1 = 1.0, 2.0
x2, y2 = 4.0, 6.0
dx = x2 - x1
dy = y2 - y1
distance = math.sqrt(dx**2 + dy**2)
print('distance:', distance)

# Min-max normalization
x, x_min, x_max = 70, 20, 120
if x_max == x_min:
    raise ValueError('x_max and x_min must differ')
x_norm = (x - x_min) / (x_max - x_min)
print('normalized:', x_norm)

# Gaussian
A, mu, sigma, x = 1.0, 0.0, 2.0, 1.0
difference = x - mu
denominator = 2 * sigma**2
exponent = -(difference**2) / denominator
G = A * math.exp(exponent)
print('G(x):', G)

# Physical area
area_pixels = 200
pixel_size_um = 0.5
area_um2 = area_pixels * pixel_size_um**2
print('area:', area_um2, 'µm²')
````

## Vectorization and masks

NumPy applies operations across arrays. A comparison produces a Boolean array. Boolean masks are fundamental in image processing because they encode statements such as **pixel intensity is greater than threshold**.

````python
array = np.array([1,2,3,4])
print(array**2)

mask = tiny_image > 100
print(mask)
bright_pixels = tiny_image[mask]
print('bright pixels:', bright_pixels)
print('mean bright intensity:', bright_pixels.mean())
````

# Part P — Real image example

Now connect Python → NumPy → image processing.

````python
import skimage as ski
image = ski.data.moon()
print('shape:', image.shape)
print('dtype:', image.dtype)
print('range:', image.min(), 'to', image.max())

plt.figure(figsize=(6,5))
plt.imshow(image, cmap='gray')
plt.title('Example grayscale image')
plt.axis('off')
plt.show()
````

````python
crop = image[150:350, 120:380]
threshold = crop.mean()
mask = crop > threshold

fig, axes = plt.subplots(1,2,figsize=(10,4))
axes[0].imshow(crop, cmap='gray')
axes[0].set_title('Crop')
axes[1].imshow(mask, cmap='gray')
axes[1].set_title('Pixels > crop mean')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()

print('crop shape:', crop.shape)
print('threshold:', threshold)
print('selected fraction:', mask.mean())
````

# Part Q — Put loops, conditions, functions, and arrays together

A reusable analysis function can combine parameter handling, conditional logic, Boolean masks, statistics, and returned results.

````python
def summarize_bright_pixels(image, threshold=None):
    """Return a bright-pixel mask and the selected mean intensity."""
    if threshold is None:
        threshold = image.mean()
    mask = image > threshold
    if mask.any():
        mean_bright = image[mask].mean()
    else:
        mean_bright = np.nan
    return mask, mean_bright

mask, mean_bright = summarize_bright_pixels(crop)
print('mean bright intensity:', mean_bright)
````
