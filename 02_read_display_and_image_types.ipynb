## Shape, channels, and data type

Before processing, inspect `shape`, `dtype`, intensity range, and whether the final axis represents color channels.

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski

def show_image(image, title='', cmap=None, figsize=(6, 5)):
    """Display one image with a clean, repeatable layout."""
    plt.figure(figsize=figsize)
    plt.imshow(image, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
gray = ski.data.camera()
color = ski.data.astronaut()

for name, image in [('gray',gray),('color',color)]:
    print(name, image.shape, image.dtype, image.min(), image.max())
```

```python
show_image(gray, 'Grayscale image', cmap='gray')
show_image(color, 'RGB image')
```

## Integer versus floating-point images

Processing functions may return floating-point images even when the original is `uint8`. Always inspect the range before casting.

```python
from skimage.util import img_as_float
gray_float = img_as_float(gray)
print('original:', gray.dtype, gray.min(), gray.max())
print('float:', gray_float.dtype, gray_float.min(), gray_float.max())
```

## Save and load a small example

```python
from pathlib import Path
import imageio.v3 as iio

Path('../outputs').mkdir(exist_ok=True)
example_crop = gray[180:330,180:330]
out_path = Path('../outputs/example_crop.png')
iio.imwrite(out_path, example_crop)
loaded = iio.imread(out_path)
print('saved to:', out_path)
print('loaded shape:', loaded.shape)
```
