## Measurements make segmentation useful

After labeling objects, `regionprops_table` can return one row per object and converts cleanly to a pandas DataFrame.

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski

def show_image(image, title='', cmap=None, figsize=(6, 5)):
    """Display one image with a clean, repeatable layout."""
    plt.figure(figsize=figsize)
    plt.imshow(image, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
from skimage import filters, measure, morphology
image = ski.data.coins()
mask = image > filters.threshold_otsu(image)
mask = morphology.remove_small_objects(mask, max_size=99)
labels = measure.label(mask)
```

## Measure area, shape, position, and intensity

```python
properties = ('label','area','centroid','eccentricity','solidity','mean_intensity')
table = measure.regionprops_table(labels, intensity_image=image, properties=properties)
df = pd.DataFrame(table)
print(df.head())
```

## Filter objects using a measurement

```python
selected = df.query('300 <= area <= 5000').copy()
print('all objects:',len(df))
print('selected objects:',len(selected))
print(selected.head())
```

## Plot a measurement distribution

```python
plt.figure(figsize=(7,4))
plt.hist(df['area'], bins=20)
plt.xlabel('Area (pixels)'); plt.ylabel('Object count'); plt.title('Object area distribution')
plt.tight_layout(); plt.show()
```

## Save a results table

```python
from pathlib import Path
Path('../outputs').mkdir(exist_ok=True)
csv_path = Path('../outputs/region_measurements.csv')
df.to_csv(csv_path, index=False)
print('saved:', csv_path)
```
