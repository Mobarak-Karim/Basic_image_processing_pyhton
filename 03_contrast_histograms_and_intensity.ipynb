## Why contrast comes before segmentation

A segmentation algorithm sees numbers, not biological meaning. Contrast inspection helps you understand what the algorithm will see.

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage as ski

def show_image(image, title='', cmap=None, figsize=(6, 5)):
    """Display one image with a clean, repeatable layout."""
    plt.figure(figsize=figsize)
    plt.imshow(image, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
image = ski.data.moon()
show_image(image, 'Original image', cmap='gray')
```

## Inspect the histogram

```python
plt.figure(figsize=(7,4))
plt.hist(image.ravel(), bins=256)
plt.xlabel('Pixel intensity')
plt.ylabel('Pixel count')
plt.title('Intensity histogram')
plt.tight_layout()
plt.show()
```

## Contrast stretching with percentiles

```python
from skimage import exposure
p2, p98 = np.percentile(image, (2,98))
stretched = exposure.rescale_intensity(image, in_range=(p2,p98))
show_image(stretched, '2–98 percentile contrast stretch', cmap='gray')
```

## Global versus local enhancement

```python
equalized = exposure.equalize_hist(image)
adaptive = exposure.equalize_adapthist(image, clip_limit=0.03)
fig, axes = plt.subplots(1,3,figsize=(14,4))
for ax,img,title in zip(axes,[image,equalized,adaptive],['Original','Histogram equalization','CLAHE / adaptive']):
    ax.imshow(img,cmap='gray'); ax.set_title(title); ax.axis('off')
plt.tight_layout(); plt.show()
```

## Practical warning
Contrast enhancement changes pixel values. Keep display enhancement separate from quantitative measurements unless it is intentionally part of the analysis.
